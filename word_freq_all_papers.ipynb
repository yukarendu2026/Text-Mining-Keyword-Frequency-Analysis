```python
# plot word cloud: 
frequencies = dict(
    zip(word_frequency_sorted["Word"], word_frequency_sorted["Frequency"])
)

from wordcloud import WordCloud
import matplotlib.pyplot as plt
import numpy as np

# Create an elliptical mask using NumPy 
height, width = 900, 1400
y, x = np.ogrid[:height, :width]
center_y, center_x = height / 2, width / 2
radius_y, radius_x = height * 0.45, width * 0.45

# Ellipse equation: (x/rx)^2 + (y/ry)^2 > 1
mask = ((x - center_x) / radius_x) ** 2 + (
    (y - center_y) / radius_y
) ** 2 > 1
mask = 255 * mask.astype(int)

font_file = "C:/Windows/Fonts/FORTE.ttf"

wordcloud = WordCloud(
    width=1200,
    height=800,
    background_color="white",
    font_path=font_file,  # Sets the custom font
    mask=mask,
    max_font_size=140,
    min_font_size=10,
    colormap='copper',
    contour_color="darkred",
    prefer_horizontal=0.6,  # 60% horizontal words, 40% vertical words
    margin = 0 # spaces between words
).generate_from_frequencies(frequencies)

plt.figure(figsize=(12, 8))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
# save word cloud figure
wordcloud.to_file("word_frequency_cloud_all_papers.png")
print("Word cloud saved!")
```

```python
# bar graph for top 20 freq words: 

top20 = word_frequency_sorted.head(20)

fig, ax = plt.subplots(figsize=(10, 8))

bars = ax.barh(top20["Word"], top20["Frequency"])

ax.invert_yaxis()
ax.set_xlabel("Frequency")
ax.set_ylabel("Keyword")
ax.set_title("Top 20 Keywords In My Publications")

# Add frequency labels
for current_bar in bars:
    width = current_bar.get_width()
    ax.text(
        width + 0.5,
        current_bar.get_y() + current_bar.get_height()/2,
        str(int(width)),
        va="center"
    )

fig.tight_layout()
plt.show()
```
